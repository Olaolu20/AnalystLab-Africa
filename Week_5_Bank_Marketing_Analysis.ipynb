```python
#impporting file
import pandas as pd
df = pd.read_csv("bank.csv")
df.head()
```

```python
#to check the numbers of rows and columns
df.shape
```

```python
#to check the data types
df.dtypes
```

```python
#to check for null values
df.isnull().sum()
```

```python
#to check the summary statistics
df.describe()
```

```python
#importing libraries for visualisation
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#to check how many customers subscribed
sns.countplot(x='deposit', data=df, hue='deposit', palette='Set2', legend=False)
plt.title('Deposit Subscription Distribution')
plt.show()
```

```python
#Age distribution
df['age'].hist(bins=30, color='steelblue', edgecolor='black')
plt.title('Age Distribution of Customers')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#Subscription rate by job type
sns.countplot(y='job', data=df, hue='deposit', palette='Set2')
plt.title('Deposit Subscription by Job Type')
plt.xlabel('Number of Customers')
plt.ylabel('Job Type')
plt.show()
```

```python
#subscrition rate by education level
sns.countplot(x='education', data=df, hue='deposit', palette='Set2')
plt.title('Deposit Subscription by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#contact method vs subscription
sns.countplot(x='contact', data=df, hue='deposit', palette='Set2')
plt.title('Deposit Subscription by Contact Method')
plt.xlabel('Contact Method')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#Previous campaign outcome vs subscription
sns.countplot(x='poutcome', data=df, hue='deposit', palette='Set2')
plt.title('Deposit Subscription by Previous Campaign Outcome')
plt.xlabel('Previous Outcome')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#correlation heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#key drivers
sns.countplot(x='housing', data=df, hue='deposit', palette='Set2')
plt.title('Deposit Subscription by Housing Loan')
plt.xlabel('Has Housing Loan')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#call duration vs subscription
sns.boxplot(x='deposit', y='duration', data=df, hue='deposit', palette='Set2')
plt.title('Call Duration by Deposit Subscription')
plt.xlabel('Subscribed to Deposit')
plt.ylabel('Call Duration (seconds)')
plt.show()
```
